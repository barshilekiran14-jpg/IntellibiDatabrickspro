```python
%sql
insert into intellibi_catlog.intelischema.CleanseMonthlysale
select distinct    
cast(trim(sale_id)As integer)as sale_id, 
upper(product) product,
upper(category)as category,
cast(trim(quantity) as integer) as quantity,
cast(trim(price) as double) as price,
to_date(trim(sale_date),'dd-MM-yyyy')as sales_date, 
upper(region)as region,
current_timestamp() as ingest_ts
from intellibi_catlog.intelischema.rawmonthlysales
where ingest_ts >
               (select coalesce(max(ingest_ts),to_date('1900-01-01','yyyy-MM-dd')) from intellibi_catlog.intelischema.CleanseMonthlysale)
order by cast(trim(sale_id)As integer)
```
