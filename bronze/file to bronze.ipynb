```python
%sql
-- USE intellibi_catlog.intelischema;
create or replace table intellibi_catlog.intelischema.rawmonthlysales
(
    sale_id string,
    product string,
    category string,
    quantity string,
    price string,
    sale_date string,
    region string,
    ingest_ts timestamp

);
```

```python
%sql
COPY INTO intellibi_catlog.intelischema.rawmonthlysales
FROM (SELECT sale_id, product, category, quantity, price,sale_date,region,
    current_timestamp() AS ingest_ts
  FROM '/Volumes/intellibi_catlog/intelischema/intelvolume/Sales_file/')
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'false');
```
