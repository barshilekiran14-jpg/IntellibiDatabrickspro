```python
 brnzTable=f"salesLake_{env}.bronze_{env}.rawmonthlysales"
 print(brnzTable)
 srcFile=f'/Volumes/saleslake_{env}/bronze_{env}/vol_saleslake_src_{env}/daily_sles/'
 print(srcFile)
```

```python
spark.sql(f'''COPY INTO {brnzTable}
FROM 
(SELECT sale_id, product, category, quantity, price, sale_date AS sales_date, region,
    current_timestamp() AS ingest_ts
  FROM '{srcFile}'
  )
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'false')
''')
```
